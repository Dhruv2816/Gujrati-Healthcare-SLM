## 3. Extract Relationships from Books (Co-occurrence)

```python
import os
import re
from src.kg.entity_extractor import extract_entities
from src.kg.neo4j_client import Neo4jClient

# Use NLTK or regex for single sentence splitting (sliding window)
sentences = [s.strip() for s in re.split(r'[\.\n]', full_corpus) if len(s.strip()) > 15]

edges_to_write = []
for i, sentence in enumerate(sentences):
    entities = extract_entities(sentence)
    
    # Build rich edges based on extracted types
    # Diseases to Symptoms
    for d in entities.diseases:
        for s in entities.symptoms:
            edges_to_write.append({
                'source': d, 'source_type': 'Disease',
                'target': s, 'target_type': 'Symptom',
                'rel_type': 'HAS_SYMPTOM', 'confidence': 0.8, 'doc_id': f'sent_{i}'
            })
    # Diseases to Treatments (Drugs vs Treatments map to TREATED_BY/MANAGED_BY inside client)
    for d in entities.diseases:
        for t in entities.drugs + entities.treatments:
            edges_to_write.append({
                'source': d, 'source_type': 'Disease',
                'target': t, 'target_type': 'Treatment',
                'confidence': 0.9, 'doc_id': f'sent_{i}'
            })

print(f"Generated {len(edges_to_write)} rich edges from sentences.")
```
